```python
import numpy as np

# Student marks
marks = np.array([78, 85, 92, 67, 74, 88, 95, 81, 59, 72])

# Calculate statistics
mean = np.mean(marks)
median = np.median(marks)
maximum = np.max(marks)
minimum = np.min(marks)
standard_deviation = np.std(marks)

# Display results
print("=" * 50)
print("       NUMPY STUDENT MARKS ANALYSIS")
print("=" * 50)

print("\nStudent Marks:")
print(marks)

print("\nStatistical Results")
print("-" * 50)

print("Mean               :", round(mean, 2))
print("Median             :", round(median, 2))
print("Maximum Mark       :", maximum)
print("Minimum Mark       :", minimum)
print("Standard Deviation :", round(standard_deviation, 2))

# Additional analysis
print("\nAdditional Analysis")
print("-" * 50)

above_average = marks[marks > mean]
below_average = marks[marks < mean]

print("Marks above mean:", above_average)
print("Marks below mean:", below_average)

print("\nShort Explanation")
print("-" * 50)
print(f"The average mark is {mean:.2f}.")
print(f"The middle mark is {median:.2f}.")
print(f"The highest mark is {maximum}.")
print(f"The lowest mark is {minimum}.")
print(
    f"The standard deviation is {standard_deviation:.2f}, "
    "which shows the spread of marks around the mean."
)
```
